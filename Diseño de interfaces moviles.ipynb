### 1. Principios esenciales del diseño móvil

* **Diseño adaptado al contexto:** El uso de dispositivos móviles se diferencia del de una computadora de escritorio porque suele realizarse durante períodos cortos y en situaciones donde la atención del usuario está dividida entre el dispositivo y su entorno. Por esta razón, el diseño debe considerar tanto la actividad que realiza la persona como las condiciones que la rodean.

* **Protección de la información:** En los dispositivos móviles es importante prevenir que el usuario pierda los datos que ha introducido. Para ello, se utilizan mecanismos como el guardado automático y otras medidas que protegen la información sin exigir acciones adicionales.

* **Principios de interacción de Norman:** El diseño debe facilitar la comprensión y el uso de los controles mediante elementos visibles, relaciones intuitivas entre las acciones y sus resultados (*mapping*), características que indiquen cómo puede utilizarse un elemento (*affordances*) y respuestas inmediatas del sistema (*feedback*), que pueden ser visuales, auditivas o táctiles.

### 2. Organización y composición: el *Wrapper*

* **El *Wrapper*:** Puede entenderse como una estructura o marco que agrupa y organiza elementos que aparecen de manera recurrente, como menús, títulos y notificaciones. Su objetivo es mantener una apariencia coherente y hacer que la navegación sea más predecible para el usuario.

* **Principales patrones de composición:**

  * **Annunciator Row:** Barra ubicada generalmente en la parte superior de la pantalla que proporciona información sobre el estado del dispositivo, la conexión y otros indicadores importantes.
  * **Notifications:** Mensajes o avisos que aparecen para informar al usuario sobre acontecimientos relevantes, procurando no interrumpir de forma innecesaria la actividad que está realizando.
  * **Fixed Menu y Revealable Menu:** Los primeros permanecen visibles y ubicados en una zona determinada de la pantalla, mientras que los segundos aparecen cuando el usuario realiza una acción, como pulsar un botón o deslizar la pantalla. Esto permite aprovechar mejor el espacio disponible.
  * **Home e Idle Screens:** Son las pantallas principales que funcionan como punto de inicio o como espacio de referencia cuando el dispositivo no está siendo utilizado activamente.

### 3. Presentación y manejo de la información

* **Formas de mostrar información:** Los contenidos pueden organizarse mediante diferentes estructuras, dependiendo de las necesidades del usuario. Entre ellas se encuentran las listas verticales, las listas de desplazamiento continuo, las cuadrículas, los carruseles y los espacios amplios como los mapas.

* **Control y confirmación de acciones:** Cuando una acción puede generar consecuencias importantes o difíciles de revertir, el sistema puede solicitar una confirmación explícita. Elementos como *Confirmation*, *Sign On* y *Cancel Protection* ayudan a reducir la posibilidad de que el usuario cometa errores accidentalmente.

* **Revelación progresiva:** Consiste en mostrar primero la información esencial y permitir que el usuario acceda posteriormente a datos más detallados. Para ello pueden utilizarse ventanas emergentes y estructuras jerárquicas, manteniendo al mismo tiempo la referencia de la pantalla desde la que se accedió.

### 4. Navegación y uso de widgets

* **Navegación lateral:** Permite desplazarse entre diferentes secciones que se encuentran al mismo nivel dentro de una aplicación. Para conseguirlo pueden utilizarse pestañas, sistemas de paginación o efectos visuales que simulan una navegación tridimensional.

* **Navegación profunda (*Drilldown*):** Organiza la información mediante relaciones jerárquicas entre elementos principales y secundarios. El usuario puede avanzar hacia niveles más específicos utilizando botones, enlaces, iconos o conjuntos de elementos organizados en forma de pila.

* **Herramientas para controlar la información:** Cuando existe una gran cantidad de contenido, se incorporan funciones como el zoom, el cambio de escala, los saltos rápidos hacia una ubicación determinada y los filtros de búsqueda dentro de una sección. Estas herramientas permiten encontrar y manipular información de manera más eficiente.

### 5. Entrada de información, gestos y sensores

* **Métodos de entrada:** Los dispositivos móviles utilizan diferentes sistemas para facilitar la introducción de información, como teclados que se adaptan a la situación, lápices digitales y escritura manual. También pueden incorporar funciones de autocompletado y predicción para reducir el tiempo necesario para escribir.

* **Gestos en pantalla (*On-screen Gestures*):** Permiten interactuar directamente con los elementos de la interfaz mediante acciones como tocar, deslizar, arrastrar o realizar movimientos con varios dedos. Un ejemplo común es el gesto de pellizcar para ampliar o reducir una imagen.

* **Gestos físicos y sensores (*Kinesthetic Gestures*):** Los dispositivos pueden aprovechar sensores como el acelerómetro o el sensor de proximidad para detectar movimientos y cambios en el entorno. De esta manera, acciones como agitar el teléfono, girarlo o modificar su orientación pueden producir determinadas respuestas del sistema.

* **Retroalimentación multisensorial:** Para informar al usuario sobre determinados acontecimientos, el dispositivo puede combinar diferentes tipos de señales, como sonidos, mensajes de voz y vibraciones. Esto resulta especialmente útil cuando la persona no está observando directamente la pantalla.
