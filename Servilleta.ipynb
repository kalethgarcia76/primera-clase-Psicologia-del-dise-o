**El incidente del “Rescate de la Servilleta”** relata una experiencia significativa de Dan Roam durante un desayuno de trabajo en Inglaterra junto a su colega Freddie. En aquella ocasión, Roam debía presentar una propuesta para desarrollar un sitio web ante un grupo de aproximadamente 50 funcionarios del gobierno. Para explicar su idea de manera sencilla, tomó varias servilletas de papel y comenzó a realizar dibujos que ayudaran a organizar y comunicar el proyecto.

Su primer esquema estaba compuesto por tres círculos relacionados entre sí: **marca, contenido y función**. Cada elemento representaba una necesidad específica del usuario: la **función** hacía referencia a las acciones que los usuarios querían realizar; el **contenido**, a la información que necesitaban conocer; y la **marca**, a aquello que la organización quería que los usuarios recordaran.

Después, Roam relacionó estos tres componentes con las diferentes áreas encargadas de desarrollar el proyecto. Los ingenieros serían responsables de crear las funciones, los redactores de desarrollar el contenido y los diseñadores de construir la experiencia asociada con la marca. Para Freddie, este sencillo dibujo permitía presentar la propuesta de una manera clara y comprensible, brindando a los directivos la confianza necesaria para apoyar la inversión.

Más adelante, durante una conferencia en la Universidad de Sheffield, Roam volvió a dibujar en una gran pizarra el esquema que había creado originalmente en las servilletas. Aunque la presentación estaba planeada para durar solo 45 minutos, la explicación generó tanto interés que terminó convirtiéndose en una conversación de casi dos horas. Finalmente, el equipo consiguió el contrato de mayor duración que había obtenido la oficina.

Esta experiencia marcó un punto importante para Roam, ya que le permitió comprender el valor de los dibujos sencillos como herramientas para organizar pensamientos, explicar problemas complejos, facilitar la toma de decisiones y transmitir ideas de manera convincente a diferentes tipos de público.
