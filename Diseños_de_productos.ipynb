El video "Leyes de UX" explora cómo la psicología cognitiva y del comportamiento sirve como cimiento fundamental para crear experiencias digitales intuitivas y centradas en el ser humano
. Su enfoque principal radica en adaptar la tecnología al funcionamiento natural de la mente humana, reduciendo la carga cognitiva y facilitando que las personas alcancen sus objetivos
.
Principios Psicología Fundamentales
Ley de Jakob: Los usuarios pasan la mayor parte de su tiempo en otros sitios y prefieren que tu producto funcione de la misma manera que los sistemas que ya conocen
. Aprovechar convenciones habituales utiliza los modelos mentales preexistentes y elimina la fricción del aprendizaje
.
Ley de Fitts: El tiempo para interactuar con un elemento depende de su tamaño y de la distancia hasta él
. Los botones y objetivos de toque deben ser suficientemente grandes y contar con el espacio adecuado entre ellos
.
Ley de Miller: La memoria de trabajo promedio retiene solo 7 (± 2) elementos a la vez
. Para no sobrecargar la capacidad mental, se utiliza el chunking o agrupación de datos en bloques más pequeños
.
Ley de Hick: El tiempo para tomar una decisión aumenta logarítmicamente con el número y la complejidad de las opciones
. Simplificar y categorizar alternativas agiliza la interacción
.
Ley de Postel (Principio de Robustez): Aconseja ser conservador en lo que el sistema genera y liberal en lo que se acepta del usuario
. Las interfaces deben tolerar entradas variadas para ofrecer una experiencia accesible y resiliente
.
Regla del Pico-Final: Las personas evalúan una experiencia basándose en cómo se sintieron en su momento más intenso (pico) y al final, en lugar del promedio de cada momento
.
Efecto de Estética-Usabilidad: Un diseño estéticamente atractivo genera una percepción positiva que lleva a los usuarios a sentir que el producto funciona mejor, llegando a enmascarar pequeños fallos de usabilidad
.
Ley de Tesler (Conservación de la Complejidad): Existe una cantidad de complejidad inherente a todo sistema que no se puede reducir
. El diseño y desarrollo deben asumir esa carga para evitar transferírsela al usuario
.
Umbral de Doherty: La productividad y la satisfacción se disparan cuando el sistema y el usuario interactúan a un ritmo inferior a 400 milisegundos
.
