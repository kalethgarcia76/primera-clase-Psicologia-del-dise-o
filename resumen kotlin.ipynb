### 1. Conceptos fundamentales y diseño del lenguaje (Capítulo 0)

* **Seguridad frente a valores nulos (*Null Safety*):** Kotlin incorpora un sistema de tipos que permite detectar posibles errores relacionados con valores nulos durante la compilación. Para trabajar de manera segura con estos valores se utilizan recursos como el operador de acceso seguro **`?.`** y el operador Elvis **`?:`**.

* **Tipos básicos del lenguaje:** Se estudian tipos fundamentales como **`Any`**, que funciona como el tipo base del que parten todos los tipos no nulos; **`Unit`**, utilizado cuando una función no necesita devolver un resultado significativo; y **`Nothing`**, empleado para representar expresiones que nunca producen un valor, como aquellas que generan una excepción o que no pueden llegar a completarse.

* **Estructuras de datos y generación de código:** Kotlin dispone de estructuras especiales como las **`data class`**, **`sealed class`** y **`value class`**. Algunas de ellas permiten que el compilador genere automáticamente métodos y funciones útiles, entre ellos **`equals()`**, **`hashCode()`**, **`toString()`** y **`componentN()`**, estos últimos utilizados principalmente para realizar desestructuración de objetos.

* **Delegación y niveles de visibilidad:** Se analizan las propiedades delegadas, incluyendo mecanismos como **`lazy`** y **`observable`**, que permiten controlar cómo se inicializan o modifican determinadas propiedades. También se diferencia entre **`lazy`** y **`lateinit`**, y se explica el funcionamiento de **`internal`**, cuyo acceso está limitado al módulo correspondiente. En la JVM, Kotlin puede utilizar técnicas como el *name mangling* para modificar los nombres en el bytecode y evitar accesos no deseados desde otros contextos.

### 2. Biblioteca estándar de Kotlin (Capítulo 1)

* **Colecciones y secuencias:** Se explica la diferencia entre las colecciones que permiten únicamente lectura y aquellas que pueden modificarse. También se estudian las **`Sequence`**, que procesan las operaciones de transformación de manera diferida. Esto permite ejecutar las operaciones cuando realmente son necesarias y evitar la creación de colecciones intermedias innecesarias.

* **Funciones de alcance (*Scope Functions*):** Se analiza el funcionamiento y la utilidad de **`let`**, **`run`**, **`with`**, **`apply`** y **`also`**. Para comprender sus diferencias se consideran aspectos como la forma en que acceden al objeto de contexto, ya sea mediante **`this`** o **`it`**, y el tipo de resultado que devuelve cada función.

### 3. Corrutinas y programación asíncrona (Capítulo 2)

* **Funcionamiento interno de las corrutinas:** Se estudia cómo el compilador transforma las funciones de suspensión mediante el modelo **Continuation-Passing Style (CPS)**. Este proceso permite representar las funciones suspendibles como máquinas de estados que utilizan la interfaz **`Continuation`** para conservar y reanudar su ejecución.

* **Herramientas para la concurrencia:** Se explican los constructores **`launch`** y **`async`**, utilizados para ejecutar tareas concurrentes con diferentes objetivos. También se aborda la función de **`CoroutineContext`** y **`CoroutineScope`** dentro del modelo de concurrencia estructurada. Además, se presentan **`Flow`**, **`StateFlow`** y **`SharedFlow`** como mecanismos para trabajar con datos que se producen y actualizan de manera reactiva.

### 4. Ecosistema de bibliotecas KotlinX (Capítulo 3)

* **Extensiones desarrolladas para Kotlin:** El ecosistema KotlinX incluye diferentes bibliotecas que amplían las capacidades del lenguaje y facilitan el desarrollo multiplataforma. Entre las principales se encuentran:

  * **`kotlinx-serialization`**: Permite convertir objetos y estructuras de datos a diferentes formatos mediante generación de código, evitando depender de la reflexión durante la ejecución.
  * **`kotlinx-datetime`**: Proporciona herramientas para trabajar con fechas, horas e intervalos de tiempo de una manera compatible con el enfoque de Kotlin.
  * **`kotlinx-collections-immutable`**: Ofrece colecciones inmutables y persistentes que permiten conservar versiones anteriores de los datos.
  * **`kotlinx-atomicfu`**: Facilita el trabajo con operaciones atómicas y concurrencia.
  * **`kotlinx-io`**: Proporciona herramientas relacionadas con operaciones de entrada y salida.
  * **`Dokka`**: Se utiliza para generar documentación a partir del código fuente.
  * **`Kover`**: Permite medir la cobertura de las pruebas sobre el código desarrollado.

### 5. Arquitectura del compilador y desarrollo de plugins (Capítulo 4)

* **Etapas del proceso de compilación:** El compilador de Kotlin puede entenderse mediante dos grandes etapas. El **Frontend** se encarga de interpretar el código, analizar su estructura y comprobar aspectos como los tipos. Posteriormente, el **Backend** transforma esa información en el formato correspondiente, como bytecode para la JVM, código JavaScript o binarios nativos.

* **Compilador K2 y representaciones intermedias:** El nuevo compilador **K2** incorpora mecanismos como **FIR (*Frontend Intermediate Representation*)** e **IR (*Intermediate Representation*)**. Estas representaciones permiten mejorar diferentes aspectos del proceso de compilación, entre ellos la velocidad y el análisis del código y de los tipos.

* **Procesamiento de anotaciones:** Se comparan tecnologías como **KAPT**, utilizada tradicionalmente para procesar anotaciones, y **KSP (*Kotlin Symbol Processing*)**, una alternativa diseñada específicamente para Kotlin que puede ofrecer un procesamiento más eficiente. También se aborda la posibilidad de desarrollar plugins personalizados que permitan ampliar las capacidades del compilador.

### 6. Kotlin Multiplatform (KMP) (Capítulo 5)

* **Compartir código entre plataformas:** **Kotlin Multiplatform** permite reutilizar una parte importante de la lógica de una aplicación entre diferentes plataformas, como Android, iOS, Web y Desktop. De esta manera, es posible compartir funcionalidades mientras se conserva la posibilidad de utilizar interfaces de usuario adaptadas a cada plataforma.

* **Sistema `expect` / `actual`:** Este mecanismo permite definir en el código común una declaración mediante **`expect`**, estableciendo lo que una funcionalidad debe ofrecer. Posteriormente, cada plataforma proporciona su propia implementación mediante **`actual`**, adaptando el comportamiento a sus características particulares.

* **Kotlin/Native e interoperabilidad:** Kotlin/Native permite convertir el código Kotlin en aplicaciones y bibliotecas nativas utilizando tecnologías como **LLVM**. También cuenta con mecanismos propios para la administración de memoria y facilita la comunicación con tecnologías del ecosistema de Apple, permitiendo interoperar con **Swift** y **Objective-C**.
